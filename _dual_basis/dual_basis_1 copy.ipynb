```python
import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go
from matrepr import mprint

from nnscaling import aesthetics
```

```python
def relu_mul(x):
    out = x * (x > 0)
    return out
```

```python
# Generate random points inside a unit disc
n_points = 3000
radii = np.sqrt(np.random.rand(n_points))  # Radii for uniform distribution in the disc
angles = np.random.rand(n_points) * 2 * np.pi  # Angles for full circle distribution

# Convert polar coordinates to Cartesian
x_coords = radii * np.cos(angles)
y_coords = radii * np.sin(angles)

# Stack coordinates to create the array of original points
original_points = np.vstack((x_coords, y_coords))

# Transformation matrix (rotation-dominant with scaling)
transformation_matrix = np.array(
    [
        [np.cos(89), -np.sin(93)],
        [np.sin(90), np.cos(90)],
    ]
)

# offset_matrix = np.array([[-0.3], [0.2]])
offset_matrix = np.array([[0], [0]])

# Apply the transformation
transformed_points = np.dot(transformation_matrix, original_points) + offset_matrix
# transformed_points = relu_mul(transformed_points)
```

Dual Basis

```python
# Compute the dual transformation matrix
dual_transformation_matrix = np.linalg.inv(transformation_matrix)

# Solution to Ax+b = 0
x_0 = -dual_transformation_matrix @ offset_matrix
# Solution after the transformation
transformed_x_0 = np.dot(transformation_matrix, x_0) + offset_matrix

# Extract dual basis vectors (columns of the inverse matrix)
dual_basis_vectors = dual_transformation_matrix
```

```python
alpha = transformation_matrix @ (original_points - x_0)
# alpha[alpha >= 0] = 0
```

```python
# Define figure
fig = go.Figure()
fig.update_layout(
    title="Points in Standard Basis and Dual Basis with Positive Orthants Highlighted",
    xaxis=dict(scaleanchor="y", title="X-Axis"),
    yaxis=dict(title="Y-Axis"),
    showlegend=True,
)


# Plot Original Points
positive_original_points = original_points[:, (original_points[0] > 0) & (original_points[1] > 0)]
fig.add_trace(
    go.Scatter(
        x=positive_original_points[0],
        y=positive_original_points[1],
        mode="markers",
        marker=dict(size=5, color="skyblue", opacity=0.5),
        name="Original Points",
    )
)
# Plot hyperplanes
for i, label in enumerate(["e1*", "e2*"]):
    # Extract the dual basis vector
    hyperplane = transformation_matrix[:, i]

    # Compute the starting and ending points
    start_point = -hyperplane  # Starting at x_0
    end_point = hyperplane  # Offset by the dual basis vector

    # Plot the vector
    fig.add_trace(
        go.Scatter(
            x=[start_point[0], end_point[0]],
            y=[start_point[1], end_point[1]],
            mode="lines+markers",
            marker=dict(size=8, color="blue"),
            line=dict(width=2, dash="dash", color="blue"),
            name=f"Hyperplane {label}",
        )
    )
# Plot transformed Points
positive_transformed_points = transformed_points[
    :, (original_points[0] > 0) & (original_points[1] > 0)
]
fig.add_trace(
    go.Scatter(
        x=positive_transformed_points[0],
        y=positive_transformed_points[1],
        mode="markers",
        marker=dict(size=5, color="grey", opacity=0.5),
        name="Original Points",
    )
)

dual_representation = np.dot(dual_basis_vectors, original_points) + x_0
positive_dual_representation = dual_representation[
    :, (original_points[0] > 0) & (original_points[1] > 0)
]
fig.add_trace(
    go.Scatter(
        x=positive_dual_representation[0, :],
        y=positive_dual_representation[1, :],
        mode="markers",
        marker=dict(size=5, color="orange", opacity=0.6),
        name="Dual Basis Representation",
    )
)


dual_representation = np.dot(dual_basis_vectors, transformed_points) + x_0
positive_dual_representation = dual_representation[
    :, (original_points[0] > 0) & (original_points[1] > 0)
]
fig.add_trace(
    go.Scatter(
        x=positive_dual_representation[0, :],
        y=positive_dual_representation[1, :],
        mode="markers",
        marker=dict(size=5, color="yellow", opacity=0.2),
        name="Dual Basis Representation",
    )
)


# Plot Dual Basis Vectors offset by x_0
for i, label in enumerate(["e1*", "e2*"]):
    # Extract the dual basis vector
    dual_vector = dual_basis_vectors[:, i]

    # Compute the starting and ending points
    start_point = x_0.flatten()  # Starting at x_0
    end_point = start_point + dual_vector  # Offset by the dual basis vector

    # Plot the vector
    fig.add_trace(
        go.Scatter(
            x=[start_point[0], end_point[0]],
            y=[start_point[1], end_point[1]],
            mode="lines+markers",
            marker=dict(size=8, color="red"),
            line=dict(width=2, dash="dash", color="red"),
            name=f"Dual Basis Vector {label}",
        )
    )

fig.show()
```

```python
from scipy.optimize import linear_sum_assignment

A = dual_basis_vectors
E = np.array([[1, 0], [0, 1]])  # Standard basis vectors e1 and e2

# Compute similarity matrix
similarities = np.dot(A.T, E) / (
    np.linalg.norm(A, axis=0).reshape(-1, 1) * np.linalg.norm(E, axis=0).reshape(1, -1)
)

# Find optimal alignment (Hungarian algorithm)
row_ind, col_ind = linear_sum_assignment(-similarities)  # Maximize similarity by negating it

# Reorder columns of A based on alignment
A_aligned = A[:, row_ind]

# Compute halfway columns
gamma = 0.5
H = gamma * A_aligned + (1 - gamma) * E[:, col_ind]

print("Aligned matrix A:\n", A_aligned)
print("Halfway matrix H:\n", H)
```
