```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import torch

from nnscaling import aesthetics
from nnscaling.data import (
    DatasetConfig,
    YinYangBinaryDataset,
    YinYangDataset,
    YinYangNoDotsBinaryDataset,
)
```

```python
fig, ax = plt.subplots()

# Define dataset
dataset_config = DatasetConfig(num_samples=2000, split="train", seed=42)
dataset = YinYangDataset(config=dataset_config)

# Figure and plot
labels = [0, 1, 2]
colors = [
    aesthetics.SeabornColors.orange,
    aesthetics.SeabornColors.blue,
    aesthetics.SeabornColors.green,
]
for i, label in enumerate(labels):
    sns.scatterplot(
        x=dataset.features[dataset.labels.squeeze() == label, 0].numpy(),
        y=dataset.features[dataset.labels.squeeze() == label, 1].numpy(),
        ax=ax,
        color=colors[i],
        marker="o",
        s=50,
    )

plt.show()
```
