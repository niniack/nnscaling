```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import torch

from nnscaling import aesthetics
from nnscaling.data import DatasetConfig, get_dataset_class
import plotly.graph_objs as go
import plotly.express as px

%load_ext autoreload
%autoreload 2
```

```python
# Define dataset
dataset_config = DatasetConfig(
    dataset_name="TorusDataset", num_samples=5000, split="train", seed=42
)
DatasetClass = get_dataset_class(name=dataset_config.dataset_name)
dataset = DatasetClass(config=dataset_config)
features, labels = dataset.features.numpy(), dataset.labels.squeeze().numpy()

# Define Seaborn colors
colors = [
    f"rgb({int(r*255)}, {int(g*255)}, {int(b*255)})"
    for r, g, b in [aesthetics.SeabornColors.blue, aesthetics.SeabornColors.orange]
]

# Generate 3D scatter plot using Plotly
fig = go.Figure(
    data=[
        go.Scatter3d(
            x=features[labels == label, 0],
            y=features[labels == label, 1],
            z=features[labels == label, 2],
            mode="markers",
            marker=dict(size=3, color=colors[label]),
            name=f"Torus {label + 1}",
        )
        for label in [0, 1]
    ]
)

# Customize layout and display
fig.update_layout(
    title="Interlocking Toruses",
    scene=dict(xaxis_title="X-axis", yaxis_title="Y-axis", zaxis_title="Z-axis"),
    margin=dict(l=0, r=0, b=0, t=40),
)
fig.show()
```

```python
fig, ax = plt.subplots()
# Define dataset
dataset_config = DatasetConfig(
    dataset_name="SunflowerDataset",
    num_samples=3_000,
    split="train",
    seed=42,
)
DatasetClass = get_dataset_class(name=dataset_config.dataset_name)
dataset = DatasetClass(config=dataset_config)

# Figure and plot
labels = [0, 1, 2]
colors = [
    aesthetics.SeabornColors.orange,
    aesthetics.SeabornColors.blue,
    aesthetics.SeabornColors.green,
]
for i, label in enumerate(labels):
    sns.scatterplot(
        x=dataset.features[dataset.labels.squeeze() == label, 0].numpy(),
        y=dataset.features[dataset.labels.squeeze() == label, 1].numpy(),
        ax=ax,
        color=colors[i],
        marker="o",
        s=50,
    )

plt.show()
```
