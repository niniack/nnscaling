```python
import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go
from matrepr import mprint

from nnscaling import aesthetics
```

```python
def relu_mul(x):
    out = x * (x > 0)
    return out
```

```python
# Generate random points inside a unit disc
n_points = 3000
radii = np.sqrt(np.random.rand(n_points))  # Radii for uniform distribution in the disc
angles = np.random.rand(n_points) * 2 * np.pi  # Angles for full circle distribution

# Convert polar coordinates to Cartesian
x_coords = radii * np.cos(angles)
y_coords = radii * np.sin(angles)

# Stack coordinates to create the array of original points
original_points = np.vstack((x_coords, y_coords))

# Transformation matrix (rotation-dominant with scaling)
transformation_matrix = np.array(
    [
        [-1.9, -0.9],
        [-0.3, 1.5],
    ]
)

# offset_matrix = np.array([[-0.3], [0.2]])
offset_matrix = np.array([[0], [0]])

# Apply the transformation
transformed_points = np.dot(transformation_matrix, original_points) + offset_matrix
transformed_points = relu_mul(transformed_points)
```

Dual Basis

```python
# Compute the dual transformation matrix
dual_transformation_matrix = np.linalg.inv(transformation_matrix)
x_0 = -dual_transformation_matrix @ offset_matrix

transformed_x_0 = np.dot(transformation_matrix, x_0) + offset_matrix

# Extract dual basis vectors (columns of the inverse matrix)
dual_basis_vectors = dual_transformation_matrix
```

```python
# Minimal Plot for Original Points in Standard Basis
fig = go.Figure()

# Plot Original Points
fig.add_trace(
    go.Scatter(
        x=original_points[0],
        y=original_points[1],
        mode="markers",
        marker=dict(size=5, color="blue", opacity=0.5),
        name="Original Points",
    )
)

# Plot Positive Orthant Points in Standard Basis
positive = (original_points[0] > 0) & (original_points[1] > 0)
print(np.sum(positive))

fig.add_trace(
    go.Scatter(
        x=original_points[0, positive],
        y=original_points[1, positive],
        mode="markers",
        marker=dict(size=5, color="cyan", opacity=0.6),
        name="Positive Orthant (Standard Basis)",
    )
)

# Plot Standard Basis Vectors
basis = np.eye(2)
for i, label in enumerate(["e1", "e2"]):
    fig.add_trace(
        go.Scatter(
            x=[0, basis[0, i]],
            y=[0, basis[1, i]],
            mode="lines+markers",
            marker=dict(size=8, color="black"),
            line=dict(width=2, dash="dash", color="black"),
            name=f"Basis Vector {label}",
        )
    )

# Plot x_0 Point
fig.add_trace(
    go.Scatter(
        x=[x_0[0, 0]],
        y=[x_0[1, 0]],
        mode="markers+text",
        marker=dict(size=12, color="black", symbol="circle"),
        text="x_0",
        textposition="top center",
        name="x_0 (Affine Origin)",
    )
)

##############

dual_representation = np.dot(dual_basis_vectors, original_points) + x_0
fig.add_trace(
    go.Scatter(
        x=dual_representation[0, :],
        y=dual_representation[1, :],
        mode="markers",
        marker=dict(size=5, color="orange", opacity=0.6),
        name="Dual Basis Representation",
    )
)

dual_representation = np.dot(dual_basis_vectors, original_points[:, positive]) + x_0
fig.add_trace(
    go.Scatter(
        x=dual_representation[0, :],
        y=dual_representation[1, :],
        mode="markers",
        marker=dict(size=5, color="yellow", opacity=0.6),
        name="Positive Orthant (Dual Basis)",
    )
)

# Plot Dual Basis Vectors offset by x_0
for i, label in enumerate(["e1*", "e2*"]):
    # Extract the dual basis vector
    dual_vector = dual_basis_vectors[:, i]

    # Compute the starting and ending points
    start_point = x_0.flatten()  # Starting at x_0
    end_point = start_point + dual_vector  # Offset by the dual basis vector

    # Plot the vector
    fig.add_trace(
        go.Scatter(
            x=[start_point[0], end_point[0]],
            y=[start_point[1], end_point[1]],
            mode="lines+markers",
            marker=dict(size=8, color="red"),
            line=dict(width=2, dash="dash", color="red"),
            name=f"Dual Basis Vector {label}",
        )
    )

# Ensure Aspect Ratio is Equal
fig.update_layout(
    title="Points in Standard Basis and Dual Basis with Positive Orthants Highlighted",
    xaxis=dict(scaleanchor="y", title="X-Axis"),
    yaxis=dict(title="Y-Axis"),
    showlegend=True,
)

fig.show()
```
